```python
# Convert Hiring Outcome Into Meaningful Labels

clean_df["hiring_status"] = clean_df["hired"].map({
    1: "Hired",
    0: "Not Hired"
})

clean_df["hiring_status"].value_counts()
```

```python
# Candidate Segmentation

clean_df["age_group"] = pd.cut(
    clean_df["age"],
    bins=[20, 24, 28, 32, 35],
    labels=["21-24", "25-28", "29-32", "33-35"]
)

clean_df["age_group"].value_counts()
```

```python
# Experience groups

clean_df["experience_group"] = pd.cut(
    clean_df["experience_years"],
    bins=[-1, 1, 3, 5, 10, 100],
    labels=[
        "0-1 Years",
        "1-3 Years",
        "3-5 Years",
        "5-10 Years",
        "10+ Years"
    ]
)

clean_df["experience_group"].value_counts()
```

```python
# CGPA Groups

clean_df["cgpa_group"] = pd.cut(
    clean_df["cgpa"],
    bins=[0, 6, 7, 8, 9, 10, 12],
    labels=[
        "Below 6",
        "6-7",
        "7-8",
        "8-9",
        "9-10",
        "10+"
    ]
)

clean_df["cgpa_group"].value_counts()
```

```python
# skills Groups

clean_df["skills_group"] = pd.cut(
    clean_df["skills_score"],
    bins=[0, 10, 20, 30, 100],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)

clean_df["skills_group"].value_counts()
```

```python
# by groups
clean_df["resume_length_group"] = pd.cut(
    clean_df["resume_length_words"],
    bins=[0, 300, 500, 700, 1000, 2000],
    labels=[
        "Short",
        "Medium",
        "Long",
        "Very Long",
        "Extremely Long"
    ]
    
)
```

```python
# Normalize CGPA

clean_df["cgpa_normalized"] = (
    clean_df["cgpa"] - clean_df["cgpa"].min()
) / (
    clean_df["cgpa"].max() - clean_df["cgpa"].min()
)
```

```python
clean_df["experience_normalized"] = (
    clean_df["experience_years"] - clean_df["experience_years"].min()
) / (
    clean_df["experience_years"].max() - clean_df["experience_years"].min()
)
```

```python
clean_df["projects_normalized"] = (
    clean_df["projects"] - clean_df["projects"].min()
) / (
    clean_df["projects"].max() - clean_df["projects"].min()
)
```

```python
clean_df["candidate_score"] = (
    clean_df["cgpa_normalized"] * 0.20 +
    clean_df["experience_normalized"] * 0.20 +
    clean_df["projects_normalized"] * 0.15 +
    clean_df["soft_skills_score"] / 10 * 0.15
)
```
