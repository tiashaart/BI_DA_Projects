```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\USER\Downloads\Data_Analysois_Projects\Sales_Business_analytics\train.csv")

df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
## 1. Monthly Sales

monthly_sales = (
    df.groupby("Year_Month")["Sales"]
    .sum()
)

monthly_sales
```

```python
## 2. Monthly Profit

monthly_profit = (
    df.groupby("Year_Month")["Sales"]
    .sum()
)

monthly_profit
```

```python
## 3. Monthly Sales Growth

monthly_growth = (
    monthly_sales
    .pct_change() * 100
)

monthly_growth
```

```python
## 4. Three-Month Rolling Average

rolling_sales = monthly_sales.rolling(3).mean()

rolling_sales
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    label="Actual Sales"
)

plt.plot(
    monthly_sales.index.astype(str),
    rolling_sales.values,
    label="3-Month Rolling Average"
)

plt.title("Sales Trend and Rolling Average")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
## 5. Year-over-Year Sales Growth

yearly_sales = df.groupby("Year")["Sales"].sum()

yearly_growth = yearly_sales.pct_change() * 100

yearly_growth
```

```python
## 6. Seasonal Sales Analysis

monthly_seasonality = (
    df.groupby("Month")["Sales"]
    .sum()
)

monthly_seasonality
```

```python
monthly_seasonality.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Sales by Month")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
## 7. Shipping Performance

df["Shipping_Days"].describe()
```
